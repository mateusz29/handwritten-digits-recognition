```python
import os
import pandas as pd
from PIL import Image
import matplotlib.pyplot as plt
import random
```

```python
image_list  = []

for num in range(10):
    path = f"dataset/{num}"
    if os.path.exists(path):
        for file_name in os.listdir(path):
            file_path = os.path.join(path, file_name)
            img = Image.open(file_path)
                
            image_list.append({
                'dir': file_path,
                'file_name': file_name,
                'image': img,
                'class': num
            })
```

```python
df = pd.DataFrame(image_list)

for i in range(10):
    group = df[df['class'] == i]
    
    plt.subplot(2, 5, i + 1)
    plt.imshow(random.choice(group['image'].tolist()))
    plt.axis(False)
    plt.title(f"Label {i}")


plt.show()
```

```python
df = pd.DataFrame(image_list)

for i in range(10):
    for x in range(10):
        group = df[df['class'] == i]
        
        plt.subplot(2, 5, x + 1)
        plt.imshow(random.choice(group['image'].tolist()), cmap='gray')
        plt.axis(False)
        plt.title(f"Sample {x}")
    plt.show()
```

```python
counts = df['class'].value_counts().sort_index()

print(counts)
print(f"Total numbers: {counts.sum()}")

plt.figure(figsize=(10, 6))
counts.plot(kind='pie', colors=plt.cm.Paired.colors, autopct='%1.1f%%')
plt.title("Distribution of classes")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import random

from tensorflow.keras.datasets import mnist
```

```python
combined_list = []

(image_list_train, label_list_train), (image_list_test, label_list_test)  = mnist.load_data()
for img, label in zip(image_list_train, label_list_train):
    combined_list.append({
        'class': label,
        'image': img
    })
for img, label in zip(image_list_test, label_list_test):
    combined_list.append({
        'class': label,
        'image': img
    })
```

```python
df = pd.DataFrame(combined_list)

for i in range(10):
    group = df[df['class'] == i]
    
    plt.subplot(2, 5, i + 1)
    plt.imshow(random.choice(group['image'].tolist()), cmap='gray')
    plt.axis(False)
    plt.title(f"Label {i}")


plt.show()
```

```python
df = pd.DataFrame(combined_list)

for i in range(10):
    for x in range(10):
        group = df[df['class'] == i]
        
        plt.subplot(2, 5, x + 1)
        plt.imshow(random.choice(group['image'].tolist()), cmap='gray')
        plt.axis(False)
        plt.title(f"Sample {x}")
    plt.show()
```

```python
counts = df['class'].value_counts().sort_index()

print(counts)
print(f"Total numbers: {counts.sum()}")

plt.figure(figsize=(10, 6))
counts.plot(kind='pie', colors=plt.cm.Paired.colors, autopct='%1.1f%%')
plt.title("Distribution of classes")
plt.show()
```
